```python
def build_preprocessor(df: pd.DataFrame, target: str) -> tuple[ColumnTransformer, list[str], list[str]]:
    """
    Input:
      - df: dataframe
      - target: columna objetivo
    Output:
      - preprocessor: ColumnTransformer
      - num_cols, cat_cols: listas de columnas
    """
    X = df.drop(columns=[target]).copy()
    num_cols = X.select_dtypes(include=np.number).columns.tolist()
    cat_cols = [c for c in X.columns if c not in num_cols]

    preprocessor = ColumnTransformer([
        ("num", Pipeline([
            ("imp", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
        ]), num_cols),
        ("cat", Pipeline([
            ("imp", SimpleImputer(strategy="most_frequent")),
            ("ohe", OneHotEncoder(handle_unknown="ignore")),
        ]), cat_cols),
    ])
    return preprocessor, num_cols, cat_cols
```

```python
def train_classifier(df: pd.DataFrame, target: str,
                     model_type: str = "logreg",
                     test_size: float = 0.2,
                     random_state: int = 42,
                     positive_value: str = "yes"):
    """
    Entrena un clasificador sklearn con pipeline (prep + modelo).
    model_type: 'logreg' o 'rf'
    """
    y_raw = df[target].astype(str).str.lower().str.strip()
    y = (y_raw == positive_value).astype(int)
    X = df.drop(columns=[target]).copy()

    preprocessor, _, _ = build_preprocessor(df, target)

    if model_type == "rf":
        model = RandomForestClassifier(n_estimators=300, random_state=random_state)
    else:
        model = LogisticRegression(max_iter=2000)

    pipe = Pipeline([("prep", preprocessor), ("model", model)])

    Xtr, Xte, ytr, yte = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )
    pipe.fit(Xtr, ytr)
    return pipe, Xtr, Xte, ytr, yte

def eval_classifier(pipe: Pipeline, Xte: pd.DataFrame, yte: pd.Series) -> dict:
    """
    Devuelve métricas principales de clasificación.
    """
    pred = pipe.predict(Xte)
    out = {"report": classification_report(yte, pred, output_dict=False)}
    if hasattr(pipe.named_steps["model"], "predict_proba"):
        proba = pipe.predict_proba(Xte)[:, 1]
        out["roc_auc"] = roc_auc_score(yte, proba)
    return out
```

```python
def train_regressor(df: pd.DataFrame, target: str,
                    model_type: str = "ridge",
                    test_size: float = 0.2,
                    random_state: int = 42,
                    time_col: str | None = None):
    """
    Entrena regresión con pipeline (prep + modelo).
    model_type: 'linear' | 'ridge' | 'lasso' | 'rf'
    Si time_col está definido, hace split temporal (train primero, test al final).
    """
    X = df.drop(columns=[target]).copy()
    y = df[target].copy()

    preprocessor, _, _ = build_preprocessor(df, target)

    if model_type == "rf":
        model = RandomForestRegressor(n_estimators=400, random_state=random_state)
    elif model_type == "lasso":
        model = Lasso(alpha=0.01)
    elif model_type == "linear":
        model = LinearRegression()
    else:
        model = Ridge(alpha=1.0)

    pipe = Pipeline([("prep", preprocessor), ("model", model)])

    if time_col and time_col in df.columns:
        dff = df.sort_values(time_col).dropna(subset=[time_col])
        X = dff.drop(columns=[target]).copy()
        y = dff[target].copy()
        split = int(len(dff) * (1 - test_size))
        Xtr, Xte = X.iloc[:split], X.iloc[split:]
        ytr, yte = y.iloc[:split], y.iloc[split:]
    else:
        Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=test_size, random_state=random_state)

    pipe.fit(Xtr, ytr)
    return pipe, Xtr, Xte, ytr, yte

def eval_regressor(pipe: Pipeline, Xte: pd.DataFrame, yte: pd.Series) -> dict:
    """
    Métricas: MAE, RMSE, R2
    """
    pred = pipe.predict(Xte)
    return {
        "MAE": mean_absolute_error(yte, pred),
        "RMSE": mean_squared_error(yte, pred),
        "R2": r2_score(yte, pred),
        "pred": pred
    }
```

```python
def get_feature_names(pipe: Pipeline) -> np.ndarray:
    """
    Devuelve nombres de features tras el preprocesado (incluye one-hot).
    """
    return pipe.named_steps["prep"].get_feature_names_out()

def get_linear_coefficients(pipe: Pipeline, top_n: int = 20) -> pd.DataFrame:
    """
    Para Linear/Ridge/Lasso/LogReg: devuelve coeficientes ordenados por |coef|.
    """
    model = pipe.named_steps["model"]
    names = get_feature_names(pipe)

    if not hasattr(model, "coef_"):
        return pd.DataFrame({"error": ["Model has no coef_ (use permutation importance)"]})

    coef = model.coef_.ravel() if model.coef_.ndim > 1 else model.coef_
    out = pd.DataFrame({"feature": names, "coef": coef})
    out["abs"] = out["coef"].abs()
    return out.sort_values("abs", ascending=False).drop(columns=["abs"]).head(top_n)

def get_permutation_importance(pipe: Pipeline, Xte: pd.DataFrame, yte: pd.Series,
                               top_n: int = 20, n_repeats: int = 10, random_state: int = 42) -> pd.DataFrame:
    """
    Importancias por permutación (sirve para RF también).
    OJO: puede tardar si hay muchas columnas.
    """
    r = permutation_importance(pipe, Xte, yte, n_repeats=n_repeats, random_state=random_state)
    names = get_feature_names(pipe)
    out = pd.DataFrame({"feature": names, "importance": r.importances_mean})
    out["abs"] = out["importance"].abs()
    return out.sort_values("abs", ascending=False).drop(columns=["abs"]).head(top_n)
```

```python
def plot_corr_heatmap(df: pd.DataFrame, num_cols: list[str] | None = None):
    """
    Heatmap de correlación (solo numéricas).
    """
    if num_cols is None:
        num_cols = df.select_dtypes(include=np.number).columns.tolist()
    if len(num_cols) < 2:
        return px.scatter(title="Not enough numeric columns for correlation heatmap")
    corr = df[num_cols].corr(numeric_only=True)
    return px.imshow(corr, title="Correlation heatmap (numeric)")
numeric_cols = ["age","chronic_count","hypertension","mental_health",
"arthritis",
"bmi",
"diabetes",
"asthma",
"cardiovascular_disease",
"copd",
"cancer_history","annual_medical_cost"]
plot_corr_heatmap(df, numeric_cols)
```
